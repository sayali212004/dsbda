```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D

# Load the Iris dataset
# Load the Iris dataset
iris_df = pd.read_csv(r"C:\Users\Sayali\Desktop\academics\dsbda_practs\iris.csv")  # Make sure the file path is correct

sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(iris_df['petal_length'], bins=20, kde=True, color='purple')
plt.title('1D Histogram of Petal Length')
plt.xlabel('Petal Length')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=iris_df, x='sepal_length', y='sepal_width', hue='species')
plt.title('2D Scatter Plot: Sepal Length vs Sepal Width')
plt.show()
```

```python
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
species_unique = iris_df['species'].unique()
colors = ['r', 'g', 'b']

for spec, color in zip(species_unique, colors):
    subset = iris_df[iris_df['species'] == spec]
    ax.scatter(
        subset['petal_length'], 
        subset['petal_width'], 
        subset['sepal_length'],
        c=color, 
        label=spec, 
        s=50
    )

ax.set_xlabel('Petal Length')
ax.set_ylabel('Petal Width')
ax.set_zlabel('Sepal Length')
ax.set_title('3D Scatter Plot')
ax.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
iris_df_sorted = iris_df.copy()
iris_df_sorted['index'] = iris_df_sorted.index
sns.lineplot(data=iris_df_sorted, x='index', y='sepal_length', hue='species')
plt.title('Simulated Temporal Plot of Sepal Length by Index')
plt.xlabel('Index (Simulated Time)')
plt.ylabel('Sepal Length')
plt.show()
```
